```python
import fanc
import fanc.plotting as fancplot
import matplotlib.pyplot as plt
import pandas as pd
import pyBigWig
```

```python
hic1 = fanc.load("../../analysis_other/porec/T2T04.hp1/hic/T2T04.hp1.hic@10000")
hic2 = fanc.load("../../analysis_other/porec/T2T04.hp2/hic/T2T04.hp2.hic@10000")

insu1= fanc.load("../../analysis_other/porec/T2T04.hp1/insulation/T2T04.hp1_25000.insulation.tsv.125000.bw")
insu2= fanc.load("../../analysis_other/porec/T2T04.hp2/insulation/T2T04.hp2_25000.insulation.tsv.125000.bw")

genes = fanc.load("../data/chm13v2.0_RefSeq_Liftoff_v5.1.sorted.gtf.gz")

boundaries = fanc.load("../outputs/pairs_files_T2T/T2T04.hp2/T2T04_boundaries_merged.bed")
H3K27ac=fanc.load("../data/chm13v2.0_HL-60.H3K27ac.bw")
```

```python

matrix = fancplot.SplitMatrixPlot(hic1, hic2,draw_ticks=False,draw_x_axis=False, vmax = 40,
                               aspect = 1,scale_matrices=False)

insu = fancplot.LinePlot((insu1,insu2),
                        fill=False,fix_chromosome=True,labels=('hp1','hp2'),
                        colors=('black','#D73C14'),ylabel = "Insulation\nScore",style="mid",
                        aspect = 0.12,
                       plot_kwargs = {"linewidth":1},draw_x_axis=False)

borders = fancplot.FeatureLayerPlot(boundaries,
                        fix_chromosome=True,aspect=0.06,ylabel="TAD\nborders",draw_x_axis=False)

gene_plot = fancplot.GenePlot(genes,group_by="gene_id",label_field = "gene_id",
                        fix_chromosome=True,squash = True,aspect=0.1, 
                        color_forward= "lightblue",color_reverse="darkblue",ylabel="genes",
                       include={"gene_id":["H19","IGF2","IGF2-AS","INS"]},
                       line_width=2.5,box_height=0.5,show_arrows=False,draw_x_axis=False)

highlight = fancplot.HighlightAnnotation([("chr11",  2087500, 2112500)],
                                   fix_chromosome=True,
                                    plot_kwargs={"alpha":0.2,
                                                "color":"black",
                                                "linewidth":1})

histone = fancplot.BarPlot(H3K27ac, fix_chromosome=True,
                           bin_size = 5000,alpha=1,
                                        ylabel = "H3K27ac",
                                        aspect = 0.07,
                          plot_kwargs={"color":"black"})
                

gf = fancplot.GenomicFigure([matrix, insu, borders, gene_plot, histone 
                             #, highlight
                            ],ticks_last=True,width=5)
                          
fig,axes = gf.plot('11:1.5mb-2.5mb')
    
cbar_ax = fig.axes[1]
cbar_ax.set_ylabel("contacts",rotation=270,labelpad=15) # add label to colorbar
cbar_ax.set_position([0.8,0.79,0.02,0.15])

matrix_ax=axes[0] # add labels to matrix
matrix_ax.text(0.9,0.95,"hp1",transform=matrix_ax.transAxes,fontsize=14,color="black")
matrix_ax.text(0.03,0.03,"hp2",transform=matrix_ax.transAxes,fontsize=14,color="black")

for spine in matrix_ax.spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1)

line_ax=axes[1] # move legend outside plot
handles,labels=line_ax.get_legend_handles_labels()
line_ax.legend(handles,labels,loc='upper left',bbox_to_anchor=(1.02,1),borderaxespad=0,frameon=False)

gene_ax=axes[3] # aes of gene names
for label in gene_ax.texts: label.set_color("black")

for ax in axes: ax.grid(False)
```

```python
fig.savefig("H19_IGF2_square.png", dpi=300)
```
