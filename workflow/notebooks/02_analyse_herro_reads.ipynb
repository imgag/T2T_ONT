## Analyse characteristics of Herro reads

```r
library(tidyverse)
library(ggplot2)
```

```r
f1 <- "/mnt/storage3b/projects/no_ngsd/ahthapp1_T2T_ONT/data/bamstats/mapped/TUE_02/TUE_02.HQ_combined.maxx_20x/bamstats.txt"
dt <- read_tsv(f1)
```

```r
dt_a <- dt %>%
    mutate(type = if_else(str_detect(name, ";"), "duplex", "herro"))

dt_a %>% head
```

```r
summary <- dt_a %>%
    group_by(type) %>%
    summarise(count = n(), across(everything(), median, na.rm = TRUE)) %>%
    select(type, count, read_length, aligned_ref_len,  match, ins, del, sub, iden, acc )
summary
```

```r
library(knitr)
library(kableExtra)

kable(summary, format = "markdown", table.attr = "style='width:100%;'") %>%
    kable_styling(full_width = FALSE, position = "left")
```
